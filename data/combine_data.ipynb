## Additional Function

```python
def merge_two_df(df1: pd.DataFrame, df2: pd.DataFrame, on: list, exclude_cols: list) -> pd.DataFrame:
    # Combining df with df_kpmm_4
    df = pd.merge(
        df1,
        df2,
        on=on,
        how='left',
        suffixes=("", "_df2")
    )

    # drop only the df2 versions of excluded columns
    drop_cols = [f"{c}_df2" for c in exclude_cols if f"{c}_df2" in df.columns]
    df = df.drop(columns=drop_cols, errors="ignore")

    return df
```

```python
# ---------- Formatting functions ----------
def needs_percent_format(series: pd.Series) -> bool:
    """Heuristic: return True if values look like proportions (-1..1)"""
    if series.dropna().empty:
        return False
    smin = series.min()
    smax = series.max()
    return (smin >= -1.0 and smax <= 1.0)

def fmt_value(x, percent_mode: bool):
    if pd.isna(x):
        return ""
    try:
        if percent_mode:
            return f"{x:.2%}"
        # else numeric with thousands separators (or float 4 decimals)
        if float(x).is_integer():
            return f"{int(x):,}"
        return f"{x:,.4f}"
    except Exception:
        return str(x)
```

```python
def generate_pivot_table_with_summaries(df_filtered: pd.DataFrame, column_to_check: str, percent_mode: bool=False) -> Tuple[pd.DataFrame, pd.DataFrame, pd.DataFrame]:
    """
    Generate pivot table with column-wise and row-wise summaries
    """

    print(f"Generating pivot table for column: {column_to_check}")

    # Create column order (sorted by year ascending, quarter order Q1..Q4)
    quarter_order = ['q1', 'q2', 'q3', 'q4']
    
    # build a sorted list of (year, quarter) present
    unique_yq = df_filtered[['year', 'quarter']].drop_duplicates()
    unique_yq['q_order'] = unique_yq['quarter'].apply(lambda q: quarter_order.index(q) if q in quarter_order else 99)
    unique_yq = unique_yq.sort_values(['year', 'q_order'])
    col_tuples = [(int(r['year']), r['quarter']) for _, r in unique_yq.iterrows()]
    
    # pivot (agg mean in case of multiple rows per company-quarter)
    pivot = df_filtered.pivot_table(index='company_name',
                               columns=['year', 'quarter'],
                               values=column_to_check,
                               aggfunc='mean')
    
    # Reindex columns to ensure sorted order and present all col_tuples
    # Build MultiIndex in desired order
    ordered_cols = pd.MultiIndex.from_tuples(col_tuples, names=['year', 'quarter'])
    # Reindex to include missing columns as NaN if needed
    pivot = pivot.reindex(columns=ordered_cols, fill_value=np.nan)
    
    # ---------- Column-wise summary (bottom row) ----------
    col_stats = {}
    series_all = pivot  # DataFrame: rows companies, cols multiindex
    for col in pivot.columns:
        s = pivot[col].dropna()
        if s.empty:
            col_stats[col] = {k: np.nan for k in ["Min","P5","P15","Q1","Mean","Median","Q3","P85","P95","Max","Std"]}
            continue
        arr = s.values
        col_stats[col] = {
            "Min": np.nanmin(arr),
            "P5": np.nanpercentile(arr, 5),
            "P15": np.nanpercentile(arr, 15),
            "Q1": np.nanpercentile(arr, 25),
            "Mean": np.nanmean(arr),
            "Median": np.nanmedian(arr),
            "Q3": np.nanpercentile(arr, 75),
            "P85": np.nanpercentile(arr, 85),
            "P95": np.nanpercentile(arr, 95),
            "Max": np.nanmax(arr),
            "Std": np.nanstd(arr, ddof=0)
        }
    
    # Build bottom summary DataFrame with same MultiIndex columns
    bottom_df = pd.DataFrame({col: pd.Series(col_stats[col]) for col in pivot.columns})
    # bottom_df index = statistic keys; columns = MultiIndex
    # We'll transpose later for display if needed.
    
    # ---------- Row-wise summary (right side) ----------
    row_stats = []
    for idx, row in pivot.iterrows():
        vals = row.dropna().values
        if vals.size == 0:
            stats_row = {k: np.nan for k in ["Min","P5","P15","Q1","Mean","Median","Q3","P85","P95","Max","Std"]}
        else:
            stats_row = {
                "Min": np.nanmin(vals),
                "P5": np.nanpercentile(vals, 5),
                "P15": np.nanpercentile(vals, 15),
                "Q1": np.nanpercentile(vals, 25),
                "Mean": np.nanmean(vals),
                "Median": np.nanmedian(vals),
                "Q3": np.nanpercentile(vals, 75),
                "P85": np.nanpercentile(vals, 85),
                "P95": np.nanpercentile(vals, 95),
                "Max": np.nanmax(vals),
                "Std": np.nanstd(vals, ddof=0)
            }
        row_stats.append((idx, stats_row))
    
    row_stats_df = pd.DataFrame([s for _, s in row_stats], index=[r for r, _ in row_stats])
    # reorder columns
    row_stats_df = row_stats_df[["Min","P5","P15","Q1","Mean","Median","Q3","P85","P95","Max","Std"]]

    # ---------- Formatting decision ----------
    # percent_mode = needs_percent_format(pivot.stack(dropna=True) if not pivot.empty else pd.Series(dtype=float))

    # Format pivot for display: create display strings but keep numeric copies for downloads if desired
    # percent_mode = True  # force percent mode for this page
    display_pivot = pivot.copy().astype(float)
    display_pivot = display_pivot.applymap(lambda x: fmt_value(x, percent_mode))

    # Create bottom summary display (as a single-row DataFrame matching pivot columns)
    display_bottom = bottom_df.applymap(lambda x: fmt_value(x, percent_mode))

    # Create row summary display
    display_row_stats = row_stats_df.applymap(lambda x: fmt_value(x, percent_mode))

    return (display_pivot, display_bottom, display_row_stats)
```

```python
def pivot_to_two_header_rows(pivot: pd.DataFrame):
    # pivot.columns is MultiIndex: (year, quarter)

    header_1 = [str(col[0]) for col in pivot.columns]   # year
    header_2 = [str(col[1]) for col in pivot.columns]   # quarter

    # Create a temporary DataFrame with 2 header rows + data
    df = pd.DataFrame([header_1, header_2] + pivot.reset_index().values.tolist())

    return df

def bottom_to_header_rows(bottom_df: pd.DataFrame):
    header_1 = [str(col[0]) for col in bottom_df.columns]
    header_2 = [str(col[1]) for col in bottom_df.columns]

    # Put statistic names as first column
    df = pd.DataFrame([ [""] + header_1,
                        ["Statistic"] + header_2 ])

    # Append the bottom summary values
    for stat_name, row in bottom_df.iterrows():
        df.loc[len(df)] = [stat_name] + row.tolist()

    return df

def row_summary_to_no_header(row_df: pd.DataFrame):
    # Row summary index = company_name
    # Columns = stats

    # Convert index to first column
    df = row_df.reset_index()

    # Create header row from column names
    header = df.columns.tolist()

    # Build dataframe: header + data rows
    final = pd.DataFrame([header] + df.values.tolist())

    return final
```

```python
def concat_all_sections(pivot, bottom_df, row_summary):
    # Convert each to header-row format
    pivot_block = pivot_to_two_header_rows(pivot)
    bottom_block = bottom_to_header_rows(bottom_df)
    row_block = row_summary_to_no_header(row_summary)

    # Add a blank row between sections (optional)
    blank = pd.DataFrame([[""] * pivot_block.shape[1]])

    # Concatenate vertically
    combined = pd.concat(
        [pivot_block, blank,
         bottom_block, blank,
         row_block],
        ignore_index=True,
        sort=False
    )

    return combined
```

```python
def _pivot_header_rows(pivot: pd.DataFrame) -> Tuple[list, list]:
    """
    Returns two lists: first_row (years), second_row (quarters) representing
    the MultiIndex columns of pivot.
    """
    # pivot.columns assumed MultiIndex (year, quarter)
    years = [str(c[0]) for c in pivot.columns]
    quarters = [str(c[1]) for c in pivot.columns]
    return years, quarters

def _row_summary_header_rows(row_stats_df: pd.DataFrame) -> Tuple[list, list]:
    """
    For the row-stats columns placed to the right, return header rows:
    - first row: blanks (same length as stats)
    - second row: the stat names (the column names of row_stats_df)
    """
    stats = list(row_stats_df.columns)
    first = [""] * len(stats)
    second = stats.copy()
    return first, second
```

```python
def assemble_right_concat(pivot: pd.DataFrame,
                          bottom_df: pd.DataFrame,
                          row_stats_df: pd.DataFrame,
                          fill_na_for_bottom: object = "") -> pd.DataFrame:
    """
    Returns a single DataFrame ready to export, where:
    - pivot columns (MultiIndex) become the first two header rows
    - row_stats_df columns are appended to the right; their first header row is blank and second header row contains stat names
    - company rows contain pivot data + row stats
    - after company rows, a blank row is inserted, then bottom summary rows (statistics x pivot-columns).
      The appended row-summary columns will contain `fill_na_for_bottom` in those bottom rows (by default "")
    """
    # --- Validate shapes / indexes
    # Ensure pivot and bottom_df share the same MultiIndex columns
    if not isinstance(pivot.columns, pd.MultiIndex):
        raise ValueError("pivot.columns must be a MultiIndex (year, quarter).")

    # order pivot columns and bottom_df columns consistently
    pivot_cols = pivot.columns
    if not (isinstance(bottom_df.columns, pd.MultiIndex) and list(bottom_df.columns) == list(pivot_cols)):
        # try reindex bottom_df columns to pivot columns if possible
        try:
            bottom_df = bottom_df.reindex(columns=pivot_cols)
        except Exception:
            raise ValueError("bottom_df columns must match pivot MultiIndex columns (year, quarter).")

    # Reindex row_stats_df rows to pivot index order if necessary
    # Keep companies order as pivot.index
    row_stats_df_aligned = row_stats_df.reindex(index=pivot.index)

    # --- Header rows
    pivot_first_row, pivot_second_row = _pivot_header_rows(pivot)
    row_first_row, row_second_row = _row_summary_header_rows(row_stats_df_aligned)

    # Combine header rows: pivot headers + row-summary headers
    header_first = pivot_first_row + row_first_row
    header_second = pivot_second_row + row_second_row

    # --- Build data section (company rows)
    # pivot values (as a 2D array) and row stats values aligned by index
    pivot_values = pivot.reset_index(drop=False)  # keep company index in leftmost column if needed
    # We want combined table where index (company_name) is first column; but user requested "no column name" - we'll
    # put the company_name as first column in the data (not in header rows). To keep consistent with earlier style,
    # we'll include the company index as the very first column in the final DataFrame content, but header rows will not include its name.
    # Extract company names
    company_names = pivot_values[pivot.index.name if pivot.index.name else 'company_name'] if pivot.index.name in pivot_values.columns else pivot.index.values
    # Actually easier: keep pivot as values and prefix an index column manually:
    pivot_data_only = pivot.copy().reset_index()  # index becomes a column (company_name)
    # Now pivot_data_only.columns = ['company_name', (year,q), ...]
    # Extract pivot numeric columns (excluding the new company_name col)
    pivot_numeric_cols = pivot.columns

    # Compose combined data rows
    combined_rows = []
    # For each company in pivot.index order
    for company in pivot.index:
        # pivot row values (in pivot column order)
        pivot_row = list(pivot.loc[company, :].values)
        # row-stats values for this company (may contain NaN)
        if company in row_stats_df_aligned.index:
            row_stats_row = list(row_stats_df_aligned.loc[company].values)
        else:
            row_stats_row = [np.nan] * len(row_second_row)
        combined_rows.append([company] + pivot_row + row_stats_row)

    # Column labels for the combined DataFrame: create a flat list
    # We will produce a DataFrame where the *very first* column is "company" (no header rows for it)
    # The user asked "no column name, I want all those 3 dataframe use their column as first row", so we will
    # produce header rows that do NOT include a name for the company column.
    combined_col_labels = [""] + [f"{h1}||{h2}" for h1, h2 in zip(pivot_first_row, pivot_second_row)] + row_second_row

    # But we want to produce multi-row header representation, not actual MultiIndex. To export as requested:
    # Build a DataFrame where first two rows are header_first/header_second (prefixed by empty cell for the company column),
    # then the company rows, then a blank row, then bottom summary rows.
    # Build the top header block
    top_header_1 = [""] + pivot_first_row + row_first_row
    top_header_2 = [""] + pivot_second_row + row_second_row

    # Build company data rows as lists (company as first element)
    data_block = combined_rows  # each row: [company, pivot... , row_stats...]

    # Build bottom summary rows: bottom_df index are stats; we will produce one row per stat
    bottom_rows = []
    bottom_index_names = list(bottom_df.index)
    for stat_name in bottom_df.index:
        # values across pivot columns in pivot order
        vals = list(bottom_df.loc[stat_name, :].values)
        # for the right side (row-stats columns), we will fill with fill_na_for_bottom (or you can compute aggregates)
        right_fill = [fill_na_for_bottom] * len(row_second_row)
        bottom_rows.append([stat_name] + vals + right_fill)

    # Now assemble full list of rows as lists of strings/numbers:
    assembled = []
    assembled.append(top_header_1)
    assembled.append(top_header_2)
    # company rows
    for r in data_block:
        assembled.append(r)
    # blank separator row
    assembled.append([""] * (1 + len(pivot_first_row) + len(row_second_row)))
    # bottom rows (statistics)
    for r in bottom_rows:
        assembled.append(r)

    # Convert to DataFrame (no column names)
    combined_df = pd.DataFrame(assembled)

    # Optionally set dtypes for numeric columns: leave as-is (pandas will use object dtype because mixed)
    return combined_df
```

```python
def add_custom_first_row(df: pd.DataFrame, text: str):
    # Create a new row with text in the first column, rest empty
    new_row = [text] + [""] * (df.shape[1] - 1)

    # Insert at the top
    df_with_header = pd.concat(
        [pd.DataFrame([new_row]), df],
        ignore_index=True
    )
    return df_with_header
```

```python
def add_empty_rows_at_bottom(df: pd.DataFrame, n: int = 2):
    # Create n empty rows with same number of columns
    empty_rows = pd.DataFrame([[""] * df.shape[1] for _ in range(n)])
    return pd.concat([df, empty_rows], ignore_index=True)
```
